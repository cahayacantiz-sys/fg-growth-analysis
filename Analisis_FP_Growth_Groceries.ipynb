```python
import pandas as pd
import matplotlib.pyplot as plt

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import fpgrowth
from mlxtend.frequent_patterns import association_rules
```

```python
import pandas as pd

df = pd.read_csv('groceries.csv', sep=';', header=None)

print(df.head())
```

## Penjelasan Dataset

Dataset ini berisi data transaksi pembelian pelanggan pada supermarket.
Data ini digunakan untuk mencari pola pembelian menggunakan algoritma FP-Growth

```python
df.shape
```

```python
transactions = []

for i in range(len(df)):
    transactions.append(df.iloc[i,0].split(','))

transactions[:5]
```

```python
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_data = te.fit(transactions).transform(transactions)

df_final = pd.DataFrame(te_data, columns=te.columns_)

df_final.head()
```

```python
from mlxtend.frequent_patterns import fpgrowth

frequent_itemsets = fpgrowth(df_final, min_support=0.02, use_colnames=True)

frequent_itemsets.head()
```

```python
from mlxtend.frequent_patterns import association_rules

rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.3)

rules.head()
```

# Analisis Pola Pembelian Menggunakan Algoritma FP-Growth

Dataset: Groceries Market Basket Dataset
Metode: FP-Growth dan Association Rules

## Penjelasan Dataset

Dataset groceries digunakan untuk menganalisis pola pembelian pelanggan disupermarket.

Dataset memiliki:
- 9835 transasksi
- 169 jenis produk

Tujuan analisis:
Mengetahui produk yang sering dibeli bersamaan menggunakan algoritma FP-Growth


## Interpretasi Dataset

Hasil `(9835, 1)` berarti:
- Dataset memiliki 9835 baris transaksi
- Memiliki 1 kolom utama yang berisi daftar barang belanja

## Interpretasi Frequent Itemsets

Frequent itemsets adalah kombinasi produk yang sering muncul bersama dalam transaksi.

Contoh:
- whole milk
- citrus fruit
- yogurt

Produk-produk tersebut memiliki nilai support tinggi sehingga sering dibeli pelanggan.

## Interpretasi Association Rules

Association rules menunjukkan hubungan antar produk yang dibeli pelanggan.

Contoh hasil:
Jika pelanggan membeli citrus fruit, maka pelanggan cenderung membeli whole milk.

Nilai confidence menunjukkan tingkat kemungkinan produk dibeli bersama.

Nilai lift > 1 menunjukkan hubungan antar produk cukup kuat.

# Kesimpulan

Berdasarkan analisis menggunakan algoritma FP-Growth, ditemukan beberapa pola pembelian pelanggan.

Produk seperti whole milk, citrus fruit, dan yogurt sering muncul bersama dalam transaksi.

Hasil association rules dapat membantu supermarket dalam:
- Penataan produk
- Strategi promosi
- Rekomendasi produk
- Analisis perilaku pelanggan
